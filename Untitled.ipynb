```python
import pandas as pd

df = pd.read_excel("data/B2B_Transaction_Data_.xlsx", dtype={"StockCode": str})
print(df.shape)
df.head()
```

```python
print(df.info())
print(df.isna().sum())
print("Duplicates:", df.duplicated().sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("NetPrice <= 0:", (df["NetPrice"] <= 0).sum())
```

```python
# Trailing spaces split identical categories (e.g. "Dünya Politika ")
# astype("string") first: some titles are numbers/dates (e.g. "1984"),
# .str.strip() would turn them into NaN otherwise
for col in ["InvoiceNo", "CustomerID", "Description", "Category", "City"]:
    df[col] = df[col].astype("string").str.strip()
```

```python
df = df.drop_duplicates()
```

```python
# Quantity NaN or 0 (price corrections) -> not sales
df = df[df["Quantity"].notna() & (df["Quantity"] > 0)]
# Price < 0.01 -> free goods / placeholder prices
df = df[(df["NetPrice"] >= 0.01) & (df["UnitPrice"] >= 0.01)]
```

```python
# Cannot be recovered from other rows -> "Unknown"
df["City"] = df["City"].fillna("Unknown")
df["Category"] = df["Category"].fillna("Unknown")
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Quantity"] = df["Quantity"].astype(int)
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Quarter"] = "Q" + df["InvoiceDate"].dt.quarter.astype(str)
df["Month"] = df["InvoiceDate"].dt.month
df["YearMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)
df["Week"] = df["InvoiceDate"].dt.isocalendar().week.astype(int)
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()
```

```python
df["Revenue"] = df["Quantity"] * df["NetPrice"]
df["GrossRevenue"] = df["Quantity"] * df["UnitPrice"]
df["DiscountRate"] = 1 - df["NetPrice"] / df["UnitPrice"]
df["InvoiceSeries"] = df["InvoiceNo"].str[:3]
```

```python
print(df.shape)   # (657855, ...) çıkmalı
print(df.isna().sum().sum())   # 0 çıkmalı
```
